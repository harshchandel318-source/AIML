```python
# Ques1 = Make a new column for Episodes count 


def extract_episodes(txt):
    start = txt.find("(", txt.rfind("("))
    end = txt.find(")", start)

    if start != -1 and end != -1:
        content = txt[start+1:end]

        if "eps" in content:
            return int(content.split()[0])

    return None
df["Episodes"] = df["Title"].apply(extract_episodes)
df
```

```python
# type conversion 

df["Episodes"].astype(int)
```

```python
#  Que2 = Make a new column for time stamps 

def extraction_time(txt):
    data = ""

    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(i + 2, min(i + 10, len(txt))):
                data += txt[j]

            return data.strip()

    return None

```

```python
df["Total time "] = df["Title"].apply(extraction_time)
df
```
