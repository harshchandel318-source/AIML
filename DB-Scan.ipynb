```python
import seaborn as sns 
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans 
import pandas as pd 
import matplotlib.pyplot as plt
import matplotlib.pyplot as plot
from sklearn.cluster import KMeans as KM
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
x = pd.DataFrame(iris.data)
x.columns = iris.feature_names
y = iris.target
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# Visualize 

sns.scatterplot(x=x_scaled[:,0] , y=x_scaled[:,2])
```

```python
# DB Scan 

from sklearn.cluster import DBSCAN
db_scan = DBSCAN(
    eps=0.8,
    min_samples=5
)
```

```python
labels = db_scan.fit_predict(x_scaled)
```

```python
sns.scatterplot(x=x_scaled[:,0] , y=x_scaled[:,2] , c=labels)
```

Non Linear Data 

```python
from sklearn.datasets import make_moons
```

```python
x , y = make_moons(
    n_samples=300 , 
    noise=0.05 ,
    random_state=42
)
```

```python
x_scalled = scaler.fit_transform(x)

sns.scatterplot(x=x_scalled[:,0] , y=x_scalled[:,1])
```

K_Means 

```python
kmeans = KMeans(
    n_clusters=2 , 
    random_state=42
)
labells = kmeans.fit_predict(x_scalled)
```

```python
sns.scatterplot(x=x_scalled[:,0] , y=x_scalled[:,1] , c=labells)
```

DB Scan

```python
db_scan = DBSCAN(
    eps=0.8,
    min_samples=5
)

label = db_scan.fit_predict(x_scalled)
```

```python
sns.scatterplot(x=x_scalled[:,0] , y=x_scalled[:,1] , c=label)
```
