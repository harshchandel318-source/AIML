```python
# Group by Aggregation 
read = {
    "category" :["A" , "B" ,"A" , "B" ,"A" , "B" ,"A" , "B" ],
    "store" :["S1" , "S1" ,"S2" , "S2", "S1" , "S2" ,"S2" , "S1"  ],
    "sales" :[100, 200, 150, 250, 120, 180, 200, 300],
    "quantity" :[10, 15, 12, 18, 8, 20, 15, 25 ],
    "Date" : pd.date_range("2023-01-01", periods=8),
}
r1 = pd.DataFrame(read)
r1
```

```python
# Group by category and calculate the sum of sales 

cat = r1.groupby("category")["sales"].sum()
cat
```

```python
# Group by store and calculate the sum of sales 

store = r1.groupby("store")["sales"].sum()
store
```

```python
# Group by multiple columns 

cat = r1.groupby(["category" , "store"])["sales"].sum()
cat
```

```python
# AGGEREGATION 

r1["sales"].mean()

# Aise hi median ,  std , variance , min , max, count ,  nikal skte h kisi bhi column ka 
```

```python
# shortcut for all operations 
r1["sales"].agg(["mean" , "median" , "sum" , "min" , "max" , "count" , "std"])

# mode print nhii kar skte h 
```

```python
# PIVOT TABLES 

tables = {
    "date" :pd.date_range("2023-01-01" , periods= 20),
    "product" :["A" , "B" , "C" , "D"] * 5,
    "region" :["east" , "west" , "north" , "sourth", "east" , "west" , "north" , "sourth", "east" , "west" ,
                "north" , "sourth", "east" , "west" , "north" , "sourth", "east" , "west" , "north" , "sourth"],

    "sales" :np.random.randint(100, 1000, 20),
    "units" :np.random.randint(10, 100, 20),
    "rep" :["john" , "mary" ,"bob" , "alice","john" , "mary" ,"bob" , "alice", "john" , "mary" 
            ,"bob" , "alice", "john" , "mary" ,"bob" , "alice", "john" , "mary" ,"bob" , "alice"]

}
t1 = pd.DataFrame(tables)

t1["Month"] = t1["date"].dt.month_name()
t1["Quarter"] = "Q" + t1["date"].dt.quarter.astype(str)

t1
```

```python
pd.pivot_table(t1, values="sales" , index="region" , columns= "product" , aggfunc="median")
```

```python
pivot2 = pd.pivot_table(t1, values=["sales" ,"units"] ,index="region" , columns= "product" )
pivot2
```

```python
# Cross Tabs 

pd.crosstab(t1 ["region"], t1 ["product"])

# yeh jo 5 aa rha h yeh no of counts h 
```
